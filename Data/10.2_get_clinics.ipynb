# Clinics

```python
import pandas as pd
import numpy as np
import os
from tqdm import tqdm
```

Get the clinics, city, department and country iterating through files in the original sivigila files

## Dengue

```python
# get denguge files .xls and xlsx
import glob
path = 'Dengue'
files = glob.glob(os.path.join(path, '*.xls')) + glob.glob(os.path.join(path, '*.xlsx'))
```

```python
# create base dataframe
df = pd.DataFrame()
columns_keep = ['Nom_upgd' ,'Pais_ocurrencia', 'Departamento_ocurrencia', 'Municipio_ocurrencia', 'Estado_final_de_caso']
# loop the files and get some columns
for file in tqdm(files):
    df_temp = pd.read_excel(file, sheet_name=0, usecols=columns_keep)
    # filter Estado_final_de_caso
    df_temp = df_temp[(df_temp['Estado_final_de_caso'] == 5) | (df_temp['Estado_final_de_caso'] == 3)]
    # append to base dataframe
    df = pd.concat([df, df_temp], ignore_index=True)
```

```python
# drop duplicates
print(len(df))
df = df.drop_duplicates()
print(len(df))
```

```python
df.head()
```

```python
df['idx'] = df['Nom_upgd'].astype(str) + '_' + df['Departamento_ocurrencia'].astype(str) + '_' + df['Municipio_ocurrencia'].astype(str)
```

```python
# get current wd
print(os.getcwd())
```

```python
df.to_csv('clinics/clinics_dengue.csv', index=False, encoding='utf-8-sig')
```

# Zika

```python
# get denguge files .xls and xlsx
path = 'Zika'
files = glob.glob(os.path.join(path, '*.xls')) + glob.glob(os.path.join(path, '*.xlsx'))

# create base dataframe
df = pd.DataFrame()
columns_keep = ['Nom_upgd' ,'Pais_ocurrencia', 'Departamento_ocurrencia', 'Municipio_ocurrencia', 'Estado_final_de_caso']
# loop the files and get some columns
for file in tqdm(files):
    df_temp = pd.read_excel(file, sheet_name=0, usecols=columns_keep)
    # filter Estado_final_de_caso
    df_temp = df_temp[(df_temp['Estado_final_de_caso'] != 2) & (df_temp['Estado_final_de_caso'] != 1)]
    # append to base dataframe
    df = pd.concat([df, df_temp], ignore_index=True)
# drop duplicates
print(len(df))
df = df.drop_duplicates()
print(len(df))
df['idx'] = df['Nom_upgd'].astype(str) + '_' + df['Departamento_ocurrencia'].astype(str) + '_' + df['Municipio_ocurrencia'].astype(str)
# get current wd
print(os.getcwd())
df.to_csv('clinics/clinics_zika.csv', index=False, encoding='utf-8-sig')
```
