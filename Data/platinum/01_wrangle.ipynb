```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('dengue_weather.csv')
df.head()
```

```python
df['date'] = pd.to_datetime(df['week'])
```

```python
df.columns
```

```python
# vars to lag
vars_to_lag = ['tavg', 'tmin', 'tmax', 'prcp', 'wdir', 'wspd', 'pres']
# group by idx_city to lag variables
for var in vars_to_lag:
    df[f'{var}_lag1'] = df.groupby('idx_city')[var].shift(1)
    
```

```python
df.head()
```

```python
df.dropna(inplace=True)  # Drop rows with NaN values after lagging
df.to_csv('dengue_weather_lagged.csv', index=False)  # Save the modified DataFrame
```
