# Elevation of Cities

```python
import requests
import pandas as pd
```

```python
# Sample DataFrame with city coordinates
df = pd.read_csv('meteo_colombia/cities_dengue.csv')
df.head()
```

```python
df.columns
```

```python
df_s = df[['idx_city', 'Latitude', 'Longitude']].copy()
```

```python
# Function to get elevation from open-elevation API
def get_elevation(lat, lon):
    url = "https://api.open-elevation.com/api/v1/lookup"
    params = {'locations': f'{lat},{lon}'}
    response = requests.get(url, params=params).json()
    
    if 'results' in response:
        return response['results'][0]['elevation']
    else:
        return None

```

```python
# Apply function to DataFrame
df['elevation'] = df.apply(lambda row: get_elevation(row['Latitude'], row['Longitude']), axis=1)

print(df)
```

```python
df_s =df[['idx_city', 'elevation']].copy()
```

```python
df_s.to_csv('meteo_colombia/cities_dengue_elevation.csv', index=False)
```
