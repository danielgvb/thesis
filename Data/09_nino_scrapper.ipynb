# Nino Index Scraper

```python
import pandas as pd 
import numpy as np
```

source: https://origin.cpc.ncep.noaa.gov/products/analysis_monitoring/ensostuff/ONI_v5.php

```python
df = pd.read_excel('silver/oni.xlsx')
df.head()
```

```python
col_names = ['year','january', 'february', 'march', 'april', 'may', 'june',
           'july', 'august', 'september', 'october', 'november', 'december']

df.columns = col_names
```

```python
# Convert year to integer
df['year'] = df['year'].astype(int)
```

```python
# Convert all other columns to float
for col in col_names[1:]:
    df[col] = df[col].astype(float)
```

```python
# wide to long dataframe
df_long = df.melt(id_vars=['year'], var_name='month', value_name='oni')
```

```python
# create date column
df_long['date'] = pd.to_datetime(df_long['year'].astype(str) + '-' + df_long['month'], format='%Y-%B')
# create year_month column
df_long['year_month'] = df_long['date'].dt.to_period('M').astype(str)
```

```python
df_final = df_long[['year_month', 'oni']].copy()
df_final.sort_values(by='year_month', inplace=True)
```

```python
df_final.to_csv('silver/oni.csv', index=False)
```
