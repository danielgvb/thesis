```python
# create dir if not exists called cties_weekly
if not os.path.exists('meteo_colombia/cities_weekly'):
    os.makedirs('meteo_colombia/cities_weekly')
```

```python
# loop over idx_city to open files in folder
path = 'meteo_colombia/cities/'
output_path = 'meteo_colombia/cities_weekly/'
cols_keep = ['time', 'tavg', 'tmin', 'tmax', 'prcp', 'wdir', 'wspd', 'pres']
cols_mean = [ 'tavg', 'tmin', 'tmax', 'prcp', 'wdir', 'wspd', 'pres']
for idx_city in df_not_empty['idx_city']:
    df_temp = pd.read_csv(path + str(idx_city) + '.csv')
    df_temp = df_temp[cols_keep]
    df_temp['date'] = pd.to_datetime(df_temp['time'])
    # get week from date as date format
    df_temp['week'] = df_temp['date'].dt.to_period('W').apply(lambda r: r.start_time)
    # group by week and get mean of cols_keep
    df_temp.drop(columns=['time'], inplace=True)
    df_temp_g = df_temp.groupby('week')[cols_mean].mean().reset_index()
    # rename week to date
    #df_temp_g.rename(columns={'week': 'date'}, inplace=True)
    # save to csv
    df_temp_g.to_csv(output_path + str(idx_city) + '.csv', index=False)
```
