# Google Trends

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import glob
```

```python
os.getcwd()
```

```python
path = 'google_trends/trends'
csv_files = glob.glob(f'{path}/*.csv')
csv_files
```

```python
df = pd.DataFrame()
for file in csv_files:
    df_test = pd.read_csv(file, skiprows=2)
    df_test.columns = ['week','google_physician', 'google_hospital', 'google_pharmacy', 'google_symptom']
    country = file.split('\\')[1].split('.')[0]
    df_test['country'] = country
    # append df_test to df, put it bellow 
    df = pd.concat([df, df_test], ignore_index=True)
    
```

```python
df['week'] = pd.to_datetime(df['week'])
```

```python
# turn week to date with to_period
import datetime as dt
# groupby week
df['week'] = df['week'].dt.to_period('W').dt.to_timestamp()
df
```

```python
df.to_csv('silver/google_trends.csv', index=False)
```
